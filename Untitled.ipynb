```python
cols_to_drop = [
    'School or Citywide Complaint',
    'Vehicle Type',
    'Taxi Company Borough',
    'Taxi Pick Up Location',
    'Garage Lot Name'
]
df.drop(columns=cols_to_drop, inplace=True)
```

```python
high_null_cols = [
    'Landmark',
    'Intersection Street 1',
    'Intersection Street 2',
    'Bridge Highway Name',
    'Bridge Highway Direction',
    'Road Ramp',
    'Bridge Highway Segment',
    'Ferry Direction',
    'Ferry Terminal Name'
]
df.drop(columns=high_null_cols, inplace=True)
```

```python
cat_cols = [
    'Descriptor', 'Location Type', 'Address Type',
    'City', 'Facility Type'
]

df[cat_cols] = df[cat_cols].fillna('Unknown')
```

```python
address_cols = [
    'Incident Address', 'Street Name',
    'Cross Street 1', 'Cross Street 2'
]

df[address_cols] = df[address_cols].fillna('Not Available')
```

```python
df['Incident Zip'] = df['Incident Zip'].fillna(df['Incident Zip'].median())
```

```python
coord_cols = [
    'X Coordinate (State Plane)',
    'Y Coordinate (State Plane)',
    'Latitude',
    'Longitude'
]

for col in coord_cols:
    df[col] = df[col].fillna(df[col].median())
```

```python
df['Created Date'] = pd.to_datetime(df['Created Date'], format='%m/%d/%Y %I:%M:%S %p', errors='coerce')
df['Closed Date'] = pd.to_datetime(df['Closed Date'], format='%m/%d/%Y %I:%M:%S %p', errors='coerce')
df['Due Date'] = pd.to_datetime(df['Due Date'], format='%m/%d/%Y %I:%M:%S %p', errors='coerce')
df['Resolution Action Updated Date'] = pd.to_datetime(
    df['Resolution Action Updated Date'],
    format='%m/%d/%Y %I:%M:%S %p',
    errors='coerce'
)
```

```python
df['Resolution Time'] = (df['Closed Date'] - df['Created Date']).dt.days
```

```python
df['Resolution Time'].fillna(df['Resolution Time'].median(), inplace=True)
```

```python
df['Closed Date'] = df['Closed Date']  # keep as is
df.loc[df['Status'] != 'Closed', 'Resolution Time'] = -1
```

```python
df['Due Date'].fillna(method='ffill', inplace=True)
df['Resolution Action Updated Date'].fillna(method='ffill', inplace=True)
```

```python
df.drop(columns=[
    'Unique Key',
    'Park Facility Name',
    'School Name', 'School Number', 'School Phone Number'
], inplace=True)
```

```python
df.loc[df['Status'] != 'Closed', 'Resolution Time'] = -1
```

```python
df.drop(columns=['Location'], inplace=True)
```

```python
df['School Region'].fillna(df['School Region'].mode()[0], inplace=True)
df['School Code'].fillna(df['School Code'].mode()[0], inplace=True)
df['School Zip'].fillna(df['School Zip'].mode()[0], inplace=True)
```

```python
from scipy.stats import kruskal

# remove invalid values
df_valid = df[df['Resolution Time'] >= 0]

# create groups
groups = [
    group['Resolution Time'].values
    for name, group in df_valid.groupby('Complaint Type')
]
```

```python
stat, p_value = kruskal(*groups)

print("Kruskal Statistic:", stat)
print("P-value:", p_value)
```

```python
if p_value > 0.05:
    print("Fail to reject H0: All sample distributions are equal")
else:
    print("Reject H0: One or more sample distributions are not equal")
```
